```python
def atm_withdrawal():
    balance = 1000.0

    try:
        amount_input = input("Enter withdrawal amount: ")
        amount = float(amount_input)

        if amount <= 0:
            print("Amount must be more than zero.")

        elif amount > balance:
            print("Insufficient balance.")

        else:
            balance -= amount
            print(f"Withdrawal successful. Remaining balance: {balance}")

    except ValueError:
        print("Please enter a valid number.")

    finally:
        print("Thank you for using the ATM.")


if __name__ == "__main__":
    atm_withdrawal()
```

```python
def book_tickets():
    available_seats = 5
    
    try:
        user_input = input("How many tickets do you want to book? ")
        tickets = int(user_input)
        
        if tickets <= 0:
            print("You must book at least 1 ticket.")
        elif tickets > available_seats:
            print("Not enough seats available.")
        else:
            print(f"Booking confirmed! You have successfully booked {tickets} ticket(s).")
            
    except ValueError:
        print("Please enter a valid number.")
        
    finally:
        print("Thank you for booking with us.")

if __name__ == "__main__":
    book_tickets()
```

```python
def add_to_cart():
    try:
        user_input = input("How many items do you want to buy? ")
        quantity = int(user_input)
        
        if quantity < 1:
            print("You must buy at least one item.")
        else:
            print(f"You added {quantity} item(s) to your cart.")
            
    except ValueError:
        print("Please enter a valid quantity.")

if __name__ == "__main__":
    add_to_cart()
```

```python
def get_day_of_week():
    days = ["", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
    
    try:
        user_input = input("Enter a number between 1 and 7: ")
        number = int(user_input)
        
        if 1 <= number <= 7:
            print(days[number])
        else:
            print("Please enter a number between 1 and 7.")
            
    except ValueError:
        print("Please enter a valid number.")

if __name__ == "__main__":
    get_day_of_week()
```

```python
def calculate_library_fine():
    fine_per_day = 0.50
    
    try:
        user_input = input("How many days late are you returning the book? ")
        days_late = float(user_input)
        
        if days_late < 0:
            print("Days late cannot be negative.")
        else:
            total_fine = days_late * fine_per_day
            print(f"Total fine: ${total_fine:.2f}")
            
    except ValueError:
        print("Please enter a valid number.")

if __name__ == "__main__":
    calculate_library_fine()
```

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def display_info(self):
        print(f"'{self.title}' by {self.author}, published in {self.year}.")

    def is_classic(self):
        print("This is a book.")

my_book = Book(title="To Kill a Mockingbird", author="Harper Lee", year=1960)
my_book.display_info()
my_book.is_classic()
```

```python
class Student:
    def __init__(self, name, grade, age):
        self.name = name
        self.grade = grade
        self.age = age

    def introduce(self):
        print(f"My name is {self.name}, I am {self.age} years old.")

    def show_grade(self):
        print(f"My grade is {self.grade}.")

student1 = Student(name="Alice", grade="A", age=16)
student1.introduce()
student1.show_grade()
```

```python
def reserve_parking_slot():
    available_slots = [101, 102, 103, 104, 105]
    reserved = [106, 107, 108, 109, 110]
    
    try:
        user_input = input("Enter a parking slot number to reserve: ")
        slot = int(user_input)
        
        if slot in available_slots:
            print(f"Slot {slot} is reserved for you.")
        elif slot in reserved:
            print(f"Slot {slot} is already reserved by someone else.")
        else:
            print(f"Slot {slot} does not exist.")
            
    except ValueError:
        print("Invalid input. Please enter a valid slot number.")

if __name__ == "__main__":
    reserve_parking_slot()
```

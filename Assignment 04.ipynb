```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]

for i, user in enumerate(users):
    if i == 3:
        print("Limit reached. Stop sending emails.")
        break
    print(f"Email sent to {user}")
```

```python
temps = [22, 25, 30, 45, 28]
critical = 40

for temp in temps:
    if temp >= critical:
        print("Critical temperature reached! Stop monitoring.")
        break
    print(f"Temperature: {temp}")
```

```python
cart = [100, 250, 50, 300]
total_price = 0

for price in cart:
    total_price += price

print(f"Total Price: {total_price}")
```

```python
customers = [
    {"name": "Ali", "paid": True},
    {"name": "Sara", "paid": False},
    {"name": "John", "paid": True}
]

for customer in customers:
    if not customer["paid"]:
        continue
    print(f"Access granted to: {customer['name']}")
```

```python
users = ["Ali", "Sara", "John"]

for user in users:
    print(f"Hello {user}, your subscription is active!")
```

```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
domains = []

for email in emails:
    domain = email.split("@")[1]
    domains.append(domain)

print(f"domains: {domains}")
```

```python
numbers = [1, 2, 3, 4, 5, 6, 9, 10, 12, 14, 15]

result = [n for n in numbers if n % 3 == 0 or n % 5 == 0]

print(result)
```

```python
numbers = [2, 3, 4]
multipliers = [1, 2, 3]

for num in numbers:
    for mult in multipliers:
        result = num * mult
        print(f"{num} x {mult} = {result}")
```
